```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('Cleaned_dataset.csv')
```

```python
df.sample(4)
```

```python
rows, columns = df.shape
print(f'Rows    : {rows}')
print(f'Columns : {columns}')
df.describe()
```

```python
y = df['Attrition'] # test data
X = df.drop(['Attrition'], axis=1) #training data

print(f'Training data: {X.columns}\n')
print(f'Testing data : {y.name}')
```

```python
y.unique()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)
```

```python
print(f'X_train : {X_train.shape}')
print(f'X_test  : {X_test.shape}\n')
print(f'y_train : {y_train.shape}')
print(f'y_test  : {y_test.shape}')
```

```python
# Number of trees in random forest
random_values = np.linspace(50, 101, 10) 
# values = []
# for i in random_values: 
#     values.append(int(i))
# n_estimators = [values]
n_estimators = [int(i) for i in random_values]

# Number of features to consider at every split
max_features = ['log2', 'sqrt']

# Maximum number of levels in tree
max_depth = [5, 10]

# Minimum number of samples required to split a node
min_samples_split = [2, 5]

# Minimum number of samples reqiured at each leaf node
min_sample_leaf = [10, 20]

# Method of selecting samples for training each tree
bootsstrap = [True, False]
```

```python
param_grid = {
    'n_estimators': n_estimators,
    'max_features': max_features,
    'max_depth': max_depth,
    'min_samples_split': min_samples_split,
    'min_samples_leaf': min_sample_leaf,
    'bootstrap': bootsstrap
}
print(param_grid)
```

```python
rfc = RandomForestClassifier(101)
```

```python
from sklearn.model_selection import GridSearchCV
rfc_grid = GridSearchCV(rfc, param_grid, cv=3, verbose=2, n_jobs=4, error_score='raise') # estimator, param_grid, cv, verbose, n_jobs
# 'verbose' gives information about the process while the fitting is happening
#  'n_jobs' use the number of cores in the cpu for this process
```

```python
rfc_grid.fit(X_train, y_train)
```

```python
rfc_grid.best_params_
```

```python
print(f'Train Accuracy: {rfc_grid.score(X_train, y_train):.3f}')
print(f'Test Accuracy: {rfc_grid.score(X_train, y_train):.3f}')
```
