```python
import pandas as pd
```

```python
df = pd.read_csv('Partially_cleaned_dataset.csv')
df.tail(6)
```

```python
att_no = (df['Attrition'] == 0)
att_yes = (df['Attrition'] == 1)
```

```python
print("Attrition(YES): ",att_yes.sum())
print("Attrition(NO) : ",att_no.sum())
```

```python
df_no = df[df['Attrition'] == 0]
df_yes = df[df['Attrition'] == 1]
```

```python
from sklearn.utils import resample
```

```python
df_downsampled_no = resample(df_no, replace=False, n_samples=199, random_state=42)
df2 = pd.concat([df_downsampled_no, df_yes])
```

```python
df2 = df2.sample(frac=1, random_state = 42).reset_index(drop=True)
```

```python
print(f"Yes: {len(df2[df2['Attrition'] == 1])}, No: {len(df2[df2['Attrition'] == 0])}")
print(f"Total records: {len(df2)}")
```

```python
df2.sample(20)
```

```python
df2.to_csv('Balanced_Set_Cross_Validation.csv')
```
