```python
# BUILD QUANTILE MODEL
def build_feature_cluster_quantile(df, feature):

    values = pd.to_numeric(df[feature], errors="coerce")

    q1 = values.quantile(0.33)
    q2 = values.quantile(0.66)

    return {"q1": q1, "q2": q2}


feature_models = {}
for feature in FEATURES_FOR_STRATEGY:
    feature_models[feature] = build_feature_cluster_quantile(df, feature)
```

```python
def get_feature_level(feature, value):

    model = feature_models[feature]
    q1, q2 = model["q1"], model["q2"]

    if value <= q1:
        return "Low"
    elif value <= q2:
        return "Medium"
    else:
        return "High"
```

```python
def recommend_strategies_for_employee(employee_features, job_role):

    strategies = set()
    cluster_summary = {}

    role_map = ROLE_BASED_STRATEGY_MAP.get(job_role)

    if role_map is None:
        raise ValueError(f"Unknown Job Role: {job_role}")

    for feature, value in employee_features.items():

        level = get_feature_level(feature, value)
        cluster_summary[feature] = level

        feature_strategy = role_map.get(feature, {})
        strategies.update(feature_strategy.get(level, []))

    return {
        "Cluster_Assignments": cluster_summary,
        "Final_Retention_Strategies": list(strategies)
    }
```

```python
def recommend_strategies_for_multiple_employees(employee_df, id_column="EmployeeID", role_column="JobRoleGroup"):

    results = {}

    for _, row in employee_df.iterrows():

        employee_id = row[id_column]
        job_role = row[role_column]

        employee_features = row.drop([id_column, role_column]).to_dict()

        results[employee_id] = recommend_strategies_for_employee(employee_features, job_role)

    return results
```

```python
employee_data = pd.DataFrame([
    {
        "EmployeeID": 101,
        "JobRoleGroup": "Clinical",
        "num__RelationshipSatisfaction": 0.4,
        "num__JobSatisfaction": 0.1,
        "OverTime": 1
    },
    {
        "EmployeeID": 102,
        "JobRoleGroup": "Technical",
        "num__RelationshipSatisfaction": 0.8,
        "num__JobSatisfaction": 0.6,
        "OverTime": 0
    }
])
```

```python
employee_results = recommend_strategies_for_multiple_employees(employee_data)

print(employee_results)
```
