```python
# Load cleaned feature dataset
feature_df = pd.read_csv("/content/drive/MyDrive/clean_retention_features_updated_newfile.csv")
print("Feature dataset loaded successfully")
```

```python
# FEATURES USED FOR STRATEGY ANALYSIS


FEATURES_FOR_STRATEGY = [

    "JobSatisfaction",
    "WorkLifeBalance",
    "MonthlyIncome",
    "JobInvolvement",
    "TrainingTimesLastYear",
    "DistanceFromHome",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "PerformanceRating",
    "YearsSinceLastPromotion",
    "YearsInCurrentRole",
    "TotalWorkingYears",
    "JobLevel",
    "OverTime"

]
```

```python
# BUILD FEATURE CLUSTER MODEL (QUANTILES)


def build_feature_cluster_quantile(df, feature):

    values = pd.to_numeric(df[feature], errors="coerce")

    q1 = values.quantile(0.33)
    q2 = values.quantile(0.66)

    return {

        "q1": float(q1),
        "q2": float(q2)

    }


feature_models = {}

for feature in FEATURES_FOR_STRATEGY:

    feature_models[feature] = build_feature_cluster_quantile(
        feature_df,
        feature
    )

print("Feature cluster models created successfully")
```

```python
content = """
import json
import os

BASE_DIR = os.path.dirname(os.path.abspath(__file__))

with open(os.path.join(BASE_DIR, 'retention_strategy_map.json'), 'r') as f:
    ROLE_BASED_STRATEGY_MAP = json.load(f)

with open(os.path.join(BASE_DIR, 'feature_cluster_model.json'), 'r') as f:
    feature_models = json.load(f)

FEATURES_FOR_STRATEGY = [
    "JobSatisfaction",
    "WorkLifeBalance",
    "MonthlyIncome",
    "JobInvolvement",
    "TrainingTimesLastYear",
    "DistanceFromHome",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "PerformanceRating",
    "YearsSinceLastPromotion",
    "YearsInCurrentRole",
    "TotalWorkingYears",
    "JobLevel",
    "OverTime"
]

def get_feature_level(feature, value):
    model = feature_models[feature]
    q1, q2 = model["q1"], model["q2"]
    if value <= q1:
        return "Low"
    elif value <= q2:
        return "Medium"
    else:
        return "High"

def recommend_strategies_for_employee(employee_id, job_role, shap_df, feature_df, top_n=2):
    emp_shap = shap_df[shap_df["EmployeeID"] == employee_id].copy()
    if emp_shap.empty:
        return None
    emp_shap["abs_shap"] = emp_shap["shap_value"].abs()
    top_features = emp_shap.sort_values("abs_shap", ascending=False).head(top_n)
    emp_features = feature_df[feature_df["EmployeeID"] == employee_id]
    if emp_features.empty:
        return None
    emp_features = emp_features.iloc[0]
    results = []
    for _, row in top_features.iterrows():
        feature = row["feature_name"]
        shap_value = row["shap_value"]
        if feature not in FEATURES_FOR_STRATEGY:
            continue
        actual_value = emp_features[feature]
        level = get_feature_level(feature, actual_value)
        role_map = ROLE_BASED_STRATEGY_MAP.get(job_role, {})
        strategy_map = role_map.get(feature, {})
        strategies = strategy_map.get(level, [])
        for strategy in strategies:
            results.append({
                "EmployeeID": employee_id,
                "Role": job_role,
                "Feature": feature,
                "FeatureValue": actual_value,
                "ClusterLevel": level,
                "SHAP_Impact": shap_value,
                "RecommendedStrategy": strategy
            })
    return results
"""

with open('strategy_engine.py', 'w') as f:
    f.write(content)

#Download the file
from google.colab import files
files.download('strategy_engine.py')
```
