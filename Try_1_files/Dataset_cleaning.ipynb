```python
import pandas as pd
```

```python
df = pd.read_csv('Feature_extracted_dataset(whd).csv')
df
```

```python
df['OverTime'] = df['OverTime'].map({'Yes': 1, 'No': 0})
```

```python
df['OverTime'].sample(5)
```

```python
df.columns
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df['MonthlyIncome'] = scaler.fit_transform(df[['MonthlyIncome']])
df['TotalWorkingYears'] = scaler.fit_transform(df[['TotalWorkingYears']])
df['DistanceFromHome'] = scaler.fit_transform(df[['DistanceFromHome']])
df['YearsAtCompany'] = scaler.fit_transform(df[['YearsAtCompany']])
df['MonthlyRate'] = scaler.fit_transform(df[['MonthlyRate']])
df['JobInvolvement'] = scaler.fit_transform(df[['JobInvolvement']])
df['JobLevel'] = scaler.fit_transform(df[['JobLevel']])
df['YearsInCurrentRole'] = scaler.fit_transform(df[['YearsInCurrentRole']])
df['EnvironmentSatisfaction'] = scaler.fit_transform(df[['EnvironmentSatisfaction']])
df['Shift'] = scaler.fit_transform(df[['Shift']])
df['WorkLifeBalance'] = scaler.fit_transform(df[['WorkLifeBalance']])
df['PercentSalaryHike'] = scaler.fit_transform(df[['PercentSalaryHike']])
```

```python
df.head(5)
```

```python
df['MaritalStatus'] = df['MaritalStatus'].map({'Single': 0, 'Married': 1, 'Divorced': 2})
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
df.sample(5)
```

```python
df['JobRole'].unique()
```

```python
df['JobRole'] = df['JobRole'].map({'Nurse': 1, 'Therapist': 2, 'Administrative': 3, 'Admin': 4, 'Other': 0})
df.sample(7)
```
