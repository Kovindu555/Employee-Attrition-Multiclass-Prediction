```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Balanced_Set_Cross_Validation.csv')
```

```python
X = df.drop(['EmployeeID', 'Attrition'], axis=1)
y = df['Attrition']
```

```python
from sklearn.model_selection import cross_val_score as cvs
```

```python
# K-Nearest Neighbors
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
cv_score_knn = cvs(knn, X, y, cv=10).mean()
print(f'Cross validation Score using KNN: {cv_score_knn}')
```

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier()
cv_score_rfc = cvs(rfc, X, y, cv=10).mean()
print(f'Cross validation Score using RFC: {cv_score_rfc}')
```

```python
# Naive Bayes
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
cv_score_gnb = cvs(gnb, X, y, cv=10).mean()
print(f'Cross validation Score using GNB: {cv_score_gnb}')
```

```python
# Support vector machine
from sklearn.svm import SVC
svm = SVC()
cv_score_svm = cvs(svm, X, y, cv=10).mean()
print(f'Cross validation Score using SVM: {cv_score_svm}')
```

```python
cv_score_list = [cv_score_knn, cv_score_rfc, cv_score_gnb, cv_score_svm]

cv_dictionary = {cv_score_knn : 'KNN', cv_score_rfc: 'RFC', cv_score_gnb: 'GNB', cv_score_svm: 'SVM'}

for score, technique in cv_dictionary.items():
    print(f'{technique} : {score:.2f}')

max_cv_value = max(cv_score_list)
print(f'\nMost suitable technique is << {cv_dictionary[max_cv_value]} >> and value is : {max_cv_value:.2f}')
```
