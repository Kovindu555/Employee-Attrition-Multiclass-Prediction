```python
def build_feature_cluster_quantile(df, feature):
    values = pd.to_numeric(df[feature], errors="coerce")
    q1 = values.quantile(0.33)
    q2 = values.quantile(0.66)
    return {"q1": q1, "q2": q2}

feature_models = {}
for feature in FEATURES_FOR_STRATEGY:
    feature_models[feature] = build_feature_cluster_quantile(feature_df, feature)
```

```python
def get_feature_level(feature, value):

    model = feature_models[feature]
    q1, q2 = model["q1"], model["q2"]

    if value <= q1:
        return "Low"
    elif value <= q2:
        return "Medium"
    else:
        return "High"
```

```python
def recommend_strategies_from_shap(
    employee_id,
    job_role,
    shap_df,
    feature_df,
    top_n=3
):
    """
    employee_id : ID of employee
    job_role    : Clinical / Technical / Administrative
    shap_df     : DataFrame containing SHAP values
    feature_df  : DataFrame containing actual feature values
    top_n       : Number of top impactful features
    """

    # Filter SHAP values for this employee
    emp_shap = shap_df[shap_df["EmployeeID"] == employee_id]

    if emp_shap.empty:
        return {"Error": "No SHAP data found for this employee"}

    # Select top N impactful features (by absolute SHAP magnitude)
    emp_shap["abs_shap"] = emp_shap["shap_value"].abs()
    top_features = emp_shap.sort_values("abs_shap", ascending=False).head(top_n)

    # Get employee's actual feature values
    emp_features = feature_df[feature_df["EmployeeID"] == employee_id]

    if emp_features.empty:
        return {"Error": "No feature data found for this employee"}

    emp_features = emp_features.iloc[0]

    strategies = []
    cluster_summary = {}

    for _, row in top_features.iterrows():

        feature_name = row["feature_name"]

        if feature_name not in FEATURES_FOR_STRATEGY:
            continue

        actual_value = emp_features[feature_name]
        level = get_feature_level(feature_name, actual_value)

        cluster_summary[feature_name] = {
            "SHAP_Value": row["shap_value"],
            "Cluster_Level": level
        }

        role_map = ROLE_BASED_STRATEGY_MAP.get(job_role, {})
        feature_strategy = role_map.get(feature_name, {})
        strategy_list = feature_strategy.get(level, [])

        for s in strategy_list:
            strategies.append({
                "Feature": feature_name,
                "SHAP_Impact": row["shap_value"],
                "Strategy": s
            })

    # Sort strategies by SHAP importance
    strategies = sorted(
        strategies,
        key=lambda x: abs(x["SHAP_Impact"]),
        reverse=True
    )

    return {
        "EmployeeID": employee_id,
        "Top_Impact_Features": cluster_summary,
        "Prioritized_Strategies": strategies
    }
```

```python
# Example SHAP dataframe from your friend
shap_df = pd.DataFrame([
    {"EmployeeID": 1313919, "feature_name": "num__JobSatisfaction", "shap_value": -0.45},
    {"EmployeeID": 1313919, "feature_name": "OverTime", "shap_value": 0.30},
    {"EmployeeID": 1313919, "feature_name": "num__MonthlyIncome", "shap_value": -0.20}
])
```

```python
result = recommend_strategies_from_shap(
    employee_id=1313919,
    job_role="Clinical",
    shap_df=shap_df,
    feature_df=feature_df,
    top_n=3
)

print(result)
```
