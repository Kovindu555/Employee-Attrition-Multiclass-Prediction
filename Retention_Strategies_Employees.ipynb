```python
def build_feature_cluster_quantile(df, feature):
    values = pd.to_numeric(df[feature], errors="coerce")
    q1 = values.quantile(0.33)
    q2 = values.quantile(0.66)
    return {"q1": q1, "q2": q2}

feature_models = {}
for feature in FEATURES_FOR_STRATEGY:
    feature_models[feature] = build_feature_cluster_quantile(feature_df, feature)
```

```python
def get_feature_level(feature, value):

    model = feature_models[feature]
    q1, q2 = model["q1"], model["q2"]

    if value <= q1:
        return "Low"
    elif value <= q2:
        return "Medium"
    else:
        return "High"
```

```python
# SINGLE EMPLOYEE STRATEGY FUNCTION

def recommend_strategies_for_employee(

        employee_id,
        job_role,
        shap_df,
        feature_df,
        top_n=2

):

    emp_shap = shap_df[
        shap_df["EmployeeID"] == employee_id
    ].copy()

    if emp_shap.empty:
        return None


    # Select most impactful features using SHAP
    emp_shap["abs_shap"] = emp_shap["shap_value"].abs()

    top_features = emp_shap.sort_values(
        "abs_shap",
        ascending=False
    ).head(top_n)


    emp_features = feature_df[
        feature_df["EmployeeID"] == employee_id
    ]

    if emp_features.empty:
        return None

    emp_features = emp_features.iloc[0]

    results = []


    for _, row in top_features.iterrows():

        feature = row["feature_name"]
        shap_value = row["shap_value"]

        if feature not in FEATURES_FOR_STRATEGY:
            continue

        actual_value = emp_features[feature]

        level = get_feature_level(
            feature,
            actual_value
        )

        role_map = ROLE_BASED_STRATEGY_MAP.get(
            job_role,
            {}
        )

        strategy_map = role_map.get(
            feature,
            {}
        )

        strategies = strategy_map.get(
            level,
            []
        )


        for strategy in strategies:

            results.append({

                "EmployeeID": employee_id,
                "Role": job_role,
                "Feature": feature,
                "FeatureValue": actual_value,
                "ClusterLevel": level,
                "SHAP_Impact": shap_value,
                "RecommendedStrategy": strategy

            })

    return results
```

```python
# MULTI EMPLOYEE ENGINE
def recommend_for_all_employees(
    shap_df,
    feature_df,
    employee_role_map,
    top_n=3
):

    all_results = []

    unique_employees = shap_df["EmployeeID"].unique()

    for emp_id in unique_employees:

        role = employee_role_map.get(emp_id, "Clinical")

        emp_result = emp_result = recommend_strategies_for_employee(
            employee_id=emp_id,
            job_role=role,
            shap_df=shap_df,
            feature_df=feature_df,
            top_n=top_n
        )

        if emp_result:
            all_results.extend(emp_result)

    return pd.DataFrame(all_results)
```

```python
# FINAL PIPELINE
def generate_retention_plan(

        risk_df,
        shap_df,
        feature_df,
        employee_role_map,
        probability_threshold=0.50,
        top_n=2

):

    high_risk_employees = risk_df[
        risk_df["RiskProbability"] >= probability_threshold
    ]["EmployeeID"]

    final_results = []


    for emp_id in high_risk_employees:

        role = employee_role_map.get(
            emp_id,
            "Clinical"
        )

        strategies = recommend_strategies_for_employee(

            employee_id=emp_id,
            job_role=role,
            shap_df=shap_df,
            feature_df=feature_df,
            top_n=top_n

        )

        if strategies:
            final_results.extend(strategies)


    return pd.DataFrame(final_results)
```

```python
risk_df = pd.DataFrame({
    "EmployeeID": feature_df["EmployeeID"],
    "RiskProbability": np.random.uniform(0.3, 0.9, len(feature_df))
})
# Create dummy SHAP explanation dataset
shap_rows = []

for emp_id in feature_df["EmployeeID"]:
    for feature in FEATURES_FOR_STRATEGY:

        shap_rows.append({
            "EmployeeID": emp_id,
            "feature_name": feature,
            "shap_value": np.random.uniform(-0.5, 0.5)
        })

shap_df = pd.DataFrame(shap_rows)

print("SHAP dataframe created")
print(shap_df.head())
# Create role mapping for employees

roles = ["Clinical", "Technical", "Administrative"]

employee_role_map = {}

for emp_id in feature_df["EmployeeID"]:

    employee_role_map[emp_id] = np.random.choice(roles)

print("Employee role map created")
```

```python

final_retention_plan = generate_retention_plan(

    risk_df=risk_df,
    shap_df=shap_df,
    feature_df=feature_df,
    employee_role_map=employee_role_map,
    probability_threshold=0.50,
    top_n=2

)


print("\nFINAL RETENTION PLAN:\n")
print(final_retention_plan)
```

```python

# EVALUATION METRICS

# COVERAGE
def calculate_coverage(risk_df, retention_plan):

    high_risk = risk_df[
        risk_df["RiskProbability"] >= 0.50
    ]["EmployeeID"].nunique()

    employees_with_strategies = retention_plan[
        "EmployeeID"
    ].nunique()

    coverage = employees_with_strategies / high_risk

    return coverage


# DIVERSITY
def calculate_diversity(retention_plan):

    total_strategies = len(retention_plan)

    unique_strategies = retention_plan[
        "RecommendedStrategy"
    ].nunique()

    diversity = unique_strategies / total_strategies

    return diversity


# AVERAGE STRATEGIES PER EMPLOYEE
def average_strategies_per_employee(retention_plan):

    total_strategies = len(retention_plan)

    total_employees = retention_plan[
        "EmployeeID"
    ].nunique()

    return total_strategies / total_employees



# RUN EVALUATION

coverage_score = calculate_coverage(risk_df, final_retention_plan)

diversity_score = calculate_diversity(final_retention_plan)

avg_strategies = average_strategies_per_employee(final_retention_plan)


print("\n==============================")
print("RETENTION STRATEGY EVALUATION")
print("==============================")

print(f"Coverage Score: {coverage_score:.2f}")

print(f"Diversity Score: {diversity_score:.2f}")

print(f"Average Strategies per Employee: {avg_strategies:.2f}")

```
