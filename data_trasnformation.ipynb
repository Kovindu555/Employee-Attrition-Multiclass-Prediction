```python
del data # For original dataset safety
```

```python
df = df.select_dtypes(include='object')
display(df.shape)
df.sample(4)
```

```python
df.drop(columns=['Over18'], inplace=True)
df.sample(6)
display(df.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False, handle_unknown='error', dtype='int64').set_output(transform='pandas')
```

### Encoding Objects to Numerical Values

#### Gender encoding (ohe) and merging

```python
enc_gender = ohe.fit_transform(df[['Gender']])
```

```python
enc_gender.sample(5)
```

```python
df = pd.concat([df, enc_gender], axis=1).drop(columns=['Gender'])
display(df.shape)
df.sample(7)
```

#### OverTime and Attrition encoding (mapping to 1s and 0s)

```python
# yes = 1 | no = 0 
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})  
df['OverTime'] = df['OverTime'].map({'Yes': 1, 'No': 0})
```
