```python
import pandas as pd
```

```python
df = pd.read_csv('Feature_extracted_dataset(whd).csv')
df.head(10)
```

```python
df = df.drop(columns=['EmployeeID'])
df
```

```python
cols = df.select_dtypes(include=["int64"]).columns
X = df[cols]
y = df["Attrition"]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained by 2 components:", pca.explained_variance_ratio_.sum())
```

```python
df_pca = pd.DataFrame(X_pca, columns=["PCA1","PCA2"])
df_out = pd.concat([df_pca, y.reset_index(drop=True)], axis=1)
df_out.head()
```
