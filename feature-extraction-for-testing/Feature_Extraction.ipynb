Feature Extraction


```python
# import statements
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
```

```python
# Reading CSV data
df = pd.read_csv('data/watson_healthcare_modified.csv')
# df.head()
```

```python
selected_columns = ['EmployeeID','Age','TotalWorkingYears','MonthlyIncome','OverTime',
                    'DistanceFromHome','YearsAtCompany','MonthlyRate',
                    'JobInvolvement','JobRole','JobLevel','YearsInCurrentRole',
                    'YearsSinceLastPromotion','YearsWithCurrManager',
                    'EnvironmentSatisfaction','Shift','WorkLifeBalance',
                    'PercentSalaryHike','MaritalStatus','Attrition']
extracted_df = df[selected_columns]
```

Check for null values and duplicates

```python
# Check for null values
print("Null values:")
print(extracted_df.isnull().sum())

# Check for duplicate values
print("\nDuplicate values:")
print(extracted_df.duplicated().sum())
```

Encode the needed values

```python
# Mapping JobRole
role_mapping = {
    'Nurse': 1,
    'Therapist': 2,
    'Administrative': 3,
    'Admin': 4,
    'Other': 0
}

extracted_df['JobRole'] = extracted_df['JobRole'].map(role_mapping)
extracted_df['JobRole']
```

```python
# Mapping OverTime
overtime_mapping = {
    'Yes': 1,
    "No": 0
}

extracted_df['OverTime'] = extracted_df['OverTime'].map(overtime_mapping)
extracted_df['OverTime']
```

```python
# Mapping Marital Status
marital_mapping = {
    'Single': 0,
    'Married': 1,
    'Divorced': 2
}

extracted_df['MaritalStatus'] = extracted_df['MaritalStatus'].map(marital_mapping)
extracted_df['MaritalStatus']
```

```python
# Mapping Attrition
attrition_mapping = {
    'Yes': 1,
    'No': 0
}

extracted_df['Attrition'] = extracted_df['Attrition'].map(attrition_mapping)
extracted_df['Attrition']
```

Normalizing

```python
scaler = MinMaxScaler()

extracted_df['MonthlyIncome'] = scaler.fit_transform(df[['MonthlyIncome']])
extracted_df['TotalWorkingYears'] = scaler.fit_transform(df[['TotalWorkingYears']])
extracted_df['DistanceFromHome'] = scaler.fit_transform(df[['DistanceFromHome']])
extracted_df['YearsAtCompany'] = scaler.fit_transform(df[['YearsAtCompany']])
extracted_df['MonthlyRate'] = scaler.fit_transform(df[['MonthlyRate']])
extracted_df['JobInvolvement'] = scaler.fit_transform(df[['JobInvolvement']])
extracted_df['JobLevel'] = scaler.fit_transform(df[['JobLevel']])
extracted_df['YearsInCurrentRole'] = scaler.fit_transform(df[['YearsInCurrentRole']])
extracted_df['EnvironmentSatisfaction'] = scaler.fit_transform(df[['EnvironmentSatisfaction']])
extracted_df['Shift'] = scaler.fit_transform(df[['Shift']])
extracted_df['WorkLifeBalance'] = scaler.fit_transform(df[['WorkLifeBalance']])
extracted_df['PercentSalaryHike'] = scaler.fit_transform(df[['PercentSalaryHike']])
```

Convert to a seperate CSV

```python
extracted_df.to_csv("data/cleaned_data.csv", index=False)
```
